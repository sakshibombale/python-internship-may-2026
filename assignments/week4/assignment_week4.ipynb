# Section A - Exception Handling

```python
"""
Q) 1. Write a function safe_divide(a, b) that divides two numbers and handles ZeroDivisionError and TypeError separately, 
printing an appropriate message for each. Use a finally block to print 'Operation attempted.' 
regardless of the outcome. Test it with valid numbers, division by zero, and a string input.
""";
```

```python
def safe_divide(a, b):
    try:
        result = a / b
        print("Result =", result)
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
    except TypeError:
        print("Error: Invalid input type")
    finally:
        print("Operation attempted.")

# Test cases
safe_divide(10, 2)      # valid numbers
safe_divide(10, 0)      # division by zero
safe_divide("10", 2)    # string input
```

```python
"""
Q) 2. Write a function validate_age(age) that raises a custom exception InvalidAgeError 
(a class you define, inheriting from Exception) if age is negative or greater than 120.
Catch this exception in a calling block and print a user-friendly message. Test it with at least 3 different values.
""";
```

```python
class InvalidAgeError(Exception):
    pass

def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError("Invalid age entered")
    else:
        print("Valid age:", age)

# Calling block
ages = [25, -5, 150]

for age in ages:
    try:
        validate_age(age)
    except InvalidAgeError:
        print("Error: Age must be between 0 and 120")
```

```python
"""
Q) 3. Write a program that repeatedly asks the user (use a loop with at least 3 simulated inputs in a list, 
e.g. user_inputs = ['10', 'abc', '0', '5']) to enter a number and divide 100 by it. 
Handle the following in a single try block using multiple except clauses:
a) ValueError if the input is not a number.
b) ZeroDivisionError if the input is zero.
c) A generic Exception as a catch-all for anything else.
Print the result if successful, otherwise print which error occurred.

""";
```

```python
user_inputs = ['10', 'abc', '0', '5']

for value in user_inputs:
    try:
        num = int(value)
        result = 100 / num
        print("Result =", result)
    except ValueError:
        print("Error: Invalid number")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
    except Exception:
        print("Error: Something went wrong")
        
```

```python
"""
Q) 4. Create a 1D NumPy array of integers from 1 to 20. Perform the following:
a) Reshape it into a 4x5 2D array.
b) Print the sum of each row and each column.
c) Extract all elements greater than 10 using boolean indexing.
d) Replace all even numbers in the array with -1.

""";
```

```python
import numpy as np

# 1D array from 1 to 20
arr = np.arange(1, 21)

# a) reshape into 4x5
arr2 = arr.reshape(4, 5)
print("4x5 Array:\n", arr2)

# b) sum of each row and column
print("Row sums:", arr2.sum(axis=1))
print("Column sums:", arr2.sum(axis=0))

# c) elements greater than 10
print("Elements > 10:", arr2[arr2 > 10])

# d) replace even numbers with -1
arr2[arr2 % 2 == 0] = -1
print("After replacing even numbers with -1:\n", arr2)
```

```python
"""
5. Create two NumPy arrays A and B as shown below and perform:
A
np.array([[1, 2], [3, 4]])
B
np.array([[5, 6], [7, 8]])
a) Element-wise addition, subtraction, and multiplication.
b) Matrix multiplication using np.dot() or the @ operator.
c) Transpose of A.
d) The determinant and inverse of A using np.linalg.

""";
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# a) Element-wise operations
print("Addition:\n", A + B)
print("Subtraction:\n", A - B)
print("Multiplication:\n", A * B)

# b) Matrix multiplication
print("Matrix Multiplication:\n", np.dot(A, B))

# c) Transpose of A
print("Transpose of A:\n", A.T)

# d) Determinant and Inverse of A
print("Determinant of A =", np.linalg.det(A))
print("Inverse of A:\n", np.linalg.inv(A))
```

```python
"""
Q) 6. Generate a NumPy array of 15 random integers between 1 and 100 using np.random.randint() (set a seed of 42 for reproducibility). Then:
a) Find the mean, median, standard deviation, min, and max.
b) Sort the array in descending order.
c) Find the indices of the 3 largest values using np.argsort().
d) Normalize the array so all values lie between 0 and 1.

""";
```

```python
import numpy as np

np.random.seed(42)
arr = np.random.randint(1, 101, 15)

print("Array:", arr)

# a) mean, median, std, min, max
print("Mean =", np.mean(arr))
print("Median =", np.median(arr))
print("Standard Deviation =", np.std(arr))
print("Minimum =", np.min(arr))
print("Maximum =", np.max(arr))

# b) sort in descending order
print("Descending order:", np.sort(arr)[::-1])

# c) indices of 3 largest values
print("Indices of 3 largest values:", np.argsort(arr)[-3:])

# d) normalize array
norm = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print("Normalized array:", norm)
```

# Section C - Pandas Basic

```python
"""
Q) 7. Create a Pandas DataFrame from the dictionary below and perform the following:
data {"Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"], "Age": [23, 21, 25, 22, 24],
      "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"], "Score": [88, 92, 79, 95, 67,
a) Display the first 3 rows and the data types of each column.
b) Add a new column Grade based on Score (>=90 'A', >=75 'B', else 'C').
c) Sort the DataFrame by Score in descending order.
d) Filter rows where Department is 'Al'.

""";
```

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],
    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}

df = pd.DataFrame(data)

# a) first 3 rows and data types
print("First 3 rows:")
print(df.head(3))
print("\nData types:")
print(df.dtypes)

# b) add Grade column
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    else:
        return 'C'

df["Grade"] = df["Score"].apply(grade)
print("\nDataFrame with Grade:")
print(df)

# c) sort by Score descending
print("\nSorted by Score:")
print(df.sort_values(by="Score", ascending=False))

# d) filter Department = AI
print("\nStudents in AI Department:")
print(df[df["Department"] == "AI"])
```

```python
"""
Q) 8. Using the same DataFrame from Q7:
a) Group by Department and find the average Score for each.
b) Find the row with the highest Score using idxmax().
c) Rename the column Score to Marks.
d) Drop the Age column and display the resulting DataFrame

""";
```

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],
    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}

df = pd.DataFrame(data)

# a) Average score of each department
print("Average Score by Department:")
print(df.groupby("Department")["Score"].mean())

# b) Row with highest score
print("\nRow with highest Score:")
print(df.loc[df["Score"].idxmax()])

# c) Rename Score to Marks
df = df.rename(columns={"Score": "Marks"})
print("\nAfter renaming Score to Marks:")
print(df)

# d) Drop Age column
df = df.drop("Age", axis=1)
print("\nAfter dropping Age column:")
print(df)
```

```python
"""
Q) 9. Create a DataFrame with some missing values as shown below and handle them:
import numpy as np data { "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
"Price": [10, np.nan, 5, 8, np.nan], "Stock": [100, 50, np.nan, 30, 201, )
a) Check the number of missing values in each column using isnull().sum().
b) Fill missing Price values with the column mean.
c) Drop rows where Stock is still missing.
d) Save the cleaned DataFrame to a CSV file named cleaned_products.csv without the index.
""";
```

```python
import pandas as pd
import numpy as np

data = {
    "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
    "Price": [10, np.nan, 5, 8, np.nan],
    "Stock": [100, 50, np.nan, 30, 20]
}

df = pd.DataFrame(data)

# a) Count missing values
print("Missing values in each column:")
print(df.isnull().sum())

# b) Fill missing Price with mean
df["Price"] = df["Price"].fillna(df["Price"].mean())

# c) Drop rows where Stock is missing
df = df.dropna(subset=["Stock"])

# d) Save cleaned DataFrame to CSV
df.to_csv("cleaned_products.csv", index=False)

print("\nCleaned DataFrame:")
print(df)
```
