# Section A - Classes and Objects

```python
"""
1. Create a class BankAccount with the following:
a) Attributes: account_holder, account_number, and balance.
b) Methods: deposit(amount), withdraw(amount), and get_balance().
c) withdraw() should print a message and do nothing if the amount exceeds the current balance.
d) A_str_method that returns a readable summary of the account.
Create two BankAccount objects and perform a few deposits and withdrawals on each.

""";
```

```python
class BankAccount:
    def __init__(self, name, acc_no, balance):
        self.name = name
        self.acc_no = acc_no
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance")
        else:
            self.balance = self.balance - amount
            print("Withdrawn:", amount)

    def get_balance(self):
        print("Balance:", self.balance)

    def __str__(self):
        return f"Name: {self.name}, Account No: {self.acc_no}, Balance: {self.balance}"

a1 = BankAccount("Sakshi", 101, 5000)
a2 = BankAccount("Riya", 102, 3000)

a1.deposit(1000)
a1.withdraw(2000)
a1.get_balance()

a2.deposit(500)
a2.withdraw(4000)
a2.get_balance()

print(a1)
print(a2)
```

```python
"""

2. Create a class Student with attributes name, roll_no, and marks (a list of 5 subject marks). Add
methods:
a) calculate_percentage() returns the average of all marks.
b) get_grade() returns 'A' for >= 80, 'B' for >= 60, 'C' for >= 40, and 'Fail' otherwise.
c) display() prints all details including percentage and grade.
Create a list of 3 Student objects and call display() on each.

""";
```

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / len(self.marks)

    def get_grade(self):
        per = self.calculate_percentage()

        if per >= 80:
            return "A"
        elif per >= 60:
            return "B"
        elif per >= 40:
            return "C"
        else:
            return "Fail"

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Marks:", self.marks)
        print("Percentage:", self.calculate_percentage())
        print("Grade:", self.get_grade())
        print()

s1 = Student("Sakshi", 1, [80, 85, 90, 75, 88])
s2 = Student("Riya", 2, [65, 70, 60, 68, 72])
s3 = Student("Pooja", 3, [35, 40, 30, 45, 38])

students = [s1, s2, s3]

for s in students:
    s.display()
```

# Section B - Encapsulation

```python
"""
3. Create a class Employee with private attributes _name, emp_id, and _salary. Add:
a) Getter and setter methods for all three attributes.
b) The salary setter should raise a ValueError if the value is below 10000.
c) A method give_raise(percent) that increases salary by the given percentage.
Demonstrate that the private attributes cannot be accessed directly, and that all access goes through the methods.

""";
```

```python
class Employee:

    def __init__(self, name, emp_id, salary):
        self._name = name
        self._emp_id = emp_id
        self._salary = salary

    def get_name(self):
        return self._name

    def get_emp_id(self):
        return self._emp_id

    def get_salary(self):
        return self._salary

    def set_name(self, name):
        self._name = name

    def set_emp_id(self, emp_id):
        self._emp_id = emp_id

    def set_salary(self, salary):
        if salary < 10000:
            raise ValueError("Salary must be at least 10000")
        self._salary = salary

    def give_raise(self, percent):
        self._salary = self._salary + (self._salary * percent / 100)


emp = Employee("Sakshi", 101, 20000)

print("Name:", emp.get_name())
print("ID:", emp.get_emp_id())
print("Salary:", emp.get_salary())

emp.give_raise(10)
print("Salary after raise:", emp.get_salary())

emp.set_salary(25000)
print("Updated Salary:", emp.get_salary())

try:
    emp.set_salary(5000)
except ValueError as e:
    print(e)
```

```python
"""
4. Build a class SecureVault that stores a private_pin and a private_data dictionary. It should have:
a) A method add_entry(pin, key, value) that adds to_data only if the pin matches.
b) A method get_entry(pin, key) that returns the value only if the pin matches.
c) Any wrong pin attempt should print 'Access Denied' and return nothing.

""";
```

```python
class SecureVault:

    def __init__(self, pin):
        self._pin = pin
        self._data = {}

    def add_entry(self, pin, key, value):
        if pin == self._pin:
            self._data[key] = value
            print("Entry Added")
        else:
            print("Access Denied")

    def get_entry(self, pin, key):
        if pin == self._pin:
            return self._data.get(key)
        else:
            print("Access Denied")


vault = SecureVault("1234")

vault.add_entry("1234", "name", "Sakshi")
vault.add_entry("1234", "city", "Pune")

print(vault.get_entry("1234", "name"))
print(vault.get_entry("1234", "city"))

print(vault.get_entry("1111", "name"))
```

# Section C- Inheritance and Method Overriding

```python
"""
5. Create a base class Shape with a method area() that returns 0 and a method describe() that prints 'I am a shape'. Create three subclasses:
a) Circle(radius) - overrides area() using pi* r^2.
b) Rectangle(length, width) - overrides area().
c) Triangle(base, height) - overrides area().
Each subclass should also override describe() to print something specific about itself. Create one
object of each and call both methods

""";
```

```python
class Shape:
    def area(self):
        return 0

    def describe(self):
        print("I am a shape")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def describe(self):
        print("I am a Circle")

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def describe(self):
        print("I am a Rectangle")

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

    def describe(self):
        print("I am a Triangle")

c = Circle(5)
r = Rectangle(4, 6)
t = Triangle(8, 5)

c.describe()
print("Area =", c.area())

r.describe()
print("Area =", r.area())

t.describe()
print("Area =", t.area())
```

# Section D - Polymorphism

```python
"""
6. Create a class Vehicle with attributes make, model, and year and a method start_engine() that prints 'Engine started'. Create two subclasses:
a) Car(make, model, year, num_doors) - overrides start_engine() to include the car's name in the message.
b) ElectricCar(make, model, year, battery_capacity) overrides start_engine() to print 'Silent start - battery at X kWh'.
Use super() in both subclasses to call the parent_init_ Create objects and test all methods

""";
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print("Engine started")

class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def start_engine(self):
        super().start_engine()
        print(f"{self.make} {self.model} car engine started")


# Child class ElectricCar
class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def start_engine(self):
        super().start_engine()
        print(f"Silent start - battery at {self.battery_capacity} kWh")


# Create objects
car1 = Car("Toyota", "Fortuner", 2023, 4)
ev1 = ElectricCar("Tesla", "Model 3", 2024, 75)

# Test methods
car1.start_engine()

print()

ev1.start_engine()
```

```python
"""
7. Create a base class Animal with a method sound() that prints 'Some sound'.
Create subclasses Dog, Cat, and Cow, each overriding sound() with their respective sounds. 
Write a function make_all_sounds(animals) that takes a list of Animal objects and calls sound() on each.
Demonstrate that the correct method is called regardless of the object type.

""";
```

```python
class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

class Cow(Animal):
    def sound(self):
        print("Moo")

def make_all_sounds(animals):
    for animal in animals:
        animal.sound()

d = Dog()
c = Cat()
w = Cow()

animals = [d, c, w]

make_all_sounds(animals)
```

```python
"""
8. Create a class Calculator with overloaded-style methods using default arguments:
a) add(a, b, c=0) - adds two or three numbers.
b) multiply(a, b, c=1) - multiplies two or three numbers.
c) Also implement_add_ on a class Vector(x, y) so that two Vector objects can be added with the + operator. Print the resulting vector.

""";
```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

    def multiply(self, a, b, c=1):
        return a * b * c

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def display(self):
        print("(", self.x, ",", self.y, ")")

c = Calculator()

print("Add 2 numbers:", c.add(10, 20))
print("Add 3 numbers:", c.add(10, 20, 30))

print("Multiply 2 numbers:", c.multiply(2, 3))
print("Multiply 3 numbers:", c.multiply(2, 3, 4))

v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2

print("Resultant Vector:")
v3.display()
```

# Section F-Menu-Driven App with OOP

```python
"""

10. Build a menu-driven Student Record Manager using OOP and file handling. The app should have a class StudentManager with 
the following menu options:
1. Add Student - takes name, roll number, and marks for 3 subjects.
2. View All displays all stored students with their average marks.
3. Search Student - finds a student by roll number.
4. Delete Student - removes a student by roll number.
5. Save to File writes all records to students_data.txt.
6. Load from File - reads records back from students_data.txt on startup.
7. Exit
Requirements: use a class with a list as internal storage, handle invalid menu choices gracefully, 
and use a while loop to keep the menu running until the user chooses to exit.

""";
```

```python
class StudentManager:
    def __init__(self):
        self.students = []
        self.load_from_file()

    def add_student(self):
        name = input("Enter Name: ")
        roll = int(input("Enter Roll Number: "))

        m1 = float(input("Enter Marks of Subject 1: "))
        m2 = float(input("Enter Marks of Subject 2: "))
        m3 = float(input("Enter Marks of Subject 3: "))

        student = [roll, name, m1, m2, m3]
        self.students.append(student)

        print("Student Added Successfully!")

    def view_all(self):
        if len(self.students) == 0:
            print("No Student Records Found!")
            return

        for s in self.students:
            avg = (s[2] + s[3] + s[4]) / 3

            print("\nRoll Number:", s[0])
            print("Name:", s[1])
            print("Average Marks:", round(avg, 2))

    def search_student(self):
        roll = int(input("Enter Roll Number to Search: "))

        for s in self.students:
            if s[0] == roll:
                avg = (s[2] + s[3] + s[4]) / 3

                print("\nStudent Found")
                print("Roll Number:", s[0])
                print("Name:", s[1])
                print("Average Marks:", round(avg, 2))
                return

        print("Student Not Found!")

    def delete_student(self):
        roll = int(input("Enter Roll Number to Delete: "))

        for s in self.students:
            if s[0] == roll:
                self.students.remove(s)
                print("Student Deleted Successfully!")
                return

        print("Student Not Found!")

    def save_to_file(self):
        file = open("students_data.txt", "w")

        for s in self.students:
            data = f"{s[0]},{s[1]},{s[2]},{s[3]},{s[4]}\n"
            file.write(data)

        file.close()
        print("Data Saved Successfully!")

    def load_from_file(self):
        try:
            file = open("students_data.txt", "r")

            for line in file:
                data = line.strip().split(",")

                student = [
                    int(data[0]),
                    data[1],
                    float(data[2]),
                    float(data[3]),
                    float(data[4])
                ]

                self.students.append(student)

            file.close()

        except FileNotFoundError:
            pass


# Main Program
sm = StudentManager()

while True:
    print("\n===== Student Record Manager =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student")
    print("4. Delete Student")
    print("5. Save To File")
    print("6. Load From File")
    print("7. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        sm.add_student()

    elif choice == "2":
        sm.view_all()

    elif choice == "3":
        sm.search_student()

    elif choice == "4":
        sm.delete_student()

    elif choice == "5":
        sm.save_to_file()

    elif choice == "6":
        sm.students = []
        sm.load_from_file()
        print("Data Loaded Successfully!")

    elif choice == "7":
        sm.save_to_file()
        print("Thank You! Exiting Program...")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```
