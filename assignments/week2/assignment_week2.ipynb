```python
"""
Q1). Write a function analyze_sentence (sentence) that takes a sentence and returns a dictionary with word_count, 
 char_count (excluding spaces), unique_words (case-insensitive), and most_frequent_word. Test it with:
 "The quick brown fox jumps over the lazy dog the fox"

""";
```

```python
def analyze_sentence(sentence):
    words = sentence.lower().split()
    result = {
        "word_count": len (words),
        "char_count": len(sentence.replace("", "")),
        "unique_words": len (set(words)),
        "most_frequent_word": max(set (words), key=words.count)
    }
    return result
text = "Python is easy and python is powerful and python is fun"
print(analyze_sentence(text))
```

```python
"""

2. Given the string below, do the following and print each result:
text = "A man a plan a canal Panama"
a) Reverse the string without using slicing.
b) Check if it is a palindrome, ignoring case and spaces.
c) Replace every vowel with *.
d) Count how many times each vowel appears.

""";
```

```python
text = "A man a plan a canal Panama"

reverse_text = ""
for ch in text:
    reverse_text = ch + reverse_text
print("Reversed String:", reverse_text)

clean_text = text.replace(" ", "").lower()

if clean_text == "".join(reversed(clean_text)):
    print("Palindrome")
else:
    print("Not Palindrome")

new_text = ""
for ch in text:
    if ch.lower() in "aeiou":
        new_text += "*"
    else:
        new_text += ch
print("Vowels Replaced:", new_text)

vowels = {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}

for ch in text.lower():
    if ch in vowels:
        vowels[ch] += 1

print("Vowel Count:", vowels)
```

```python
"""
3. Write a function format_report(data) that takes a list of tuples (name, score, grade) and prints 
a formatted table where name is left-aligned (width 15), score is right-aligned (width 6), and grade is centered (width 8). Use this data:
data - [("Alice", 92, "A"), ("Bob", 78, "B+"), ("Charlie", 55, "C"), ("Diana", 100, "ΛΙ")]

""";
```

```python
def format_report(data):
    print(f"{'Name':<15}{'Score':>6}{'Grade':^8}")

    for name, score, grade in data:
        print(f"{name:<15}{score:>6}{grade:^8}")


data = [
    ("sakshi", 92, "A"),
    ("bhargavi", 78, "B+"),
    ("yashshri", 55, "C"),
    ("aditi", 100, "A+")
]

format_report(data)
```

```python

"""
4. Write a function list_ops(nums) that:
a) Removes duplicates while preserving the original order.
b) Returns a new list with only elements at even indices, squared.
c) Flattens the nested list into a single list using list comprehension.
nums [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5] nested [[1, 2, 3], [4, 5], [6, 7, 8,
9]]

""";
```

```python
def list_ops(nums, nested):

    unique = []
    for num in nums:
        if num not in unique:
            unique.append(num)

    squared = [nums[i]**2 for i in range(0, len(nums), 2)]

    flat = [item for sublist in nested for item in sublist]

    print("Without duplicates:", unique)
    print("Even index squared:", squared)
    print("Flattened list:", flat)


nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
nested = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]

list_ops(nums, nested)
```

```python
"""
7. Write a function inventory_manager(records) that takes a list of tuples (product, category, quantity, price)
and returns a dictionary grouped by category. Each category should have total_items, total_value, and a list of product names. Use:
records [ ("Laptop", "Electronics", 10, 55000), ("Mouse", "Electronics", 50, 499),
("Desk", "Furniture", 8, 12000), ("Chair", "Furnilure", 15, 4500), ("Pen", "Stationery", 200, 20),

""";
```

```python
def inventory_manager(records):
    result = {}

    for product, category, quantity, price in records:

        if category not in result:
            result[category] = {
                "total_items": 0,
                "total_value": 0,
                "products": []
            }

        result[category]["total_items"] += quantity
        result[category]["total_value"] += quantity * price
        result[category]["products"].append(product)

    return result


records = [
    ("Laptop", "Electronics", 10, 55000),
    ("Mouse", "Electronics", 50, 499),
    ("Desk", "Furniture", 8, 12000),
    ("Chair", "Furniture", 15, 4500),
    ("Pen", "Stationery", 200, 20)
]

print(inventory_manager(records))
```

```python
"""
8. Implement the following:
a) A function power_all(*args, exp=2) that returns a list of each number raised to exp.
b) A function profile(**kwargs) that prints each key-value pair as Key: Value.
c) Use a lambda inside sorted() to sort the list of dicts below by cgpa in descending order.
students = [{"name": "Arjun", "cgpa": 8.4}, {"name": "Sanya", "egpa": 9.1), {"name": "Dev", "cgpa": 7.8)]

""";
```

```python
def power_all(*args, exp=2):
    result = []

    for i in args:
        result.append(i ** exp)

    return result

print(power_all(2, 3, 4))
print(power_all(2, 3, 4, exp=3))

def profile(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)


profile(name="Sakshi", age=18, city="Bhiwandi")

students = [
    {"name": "Arjun", "cgpa": 8.4},
    {"name": "Sanya", "cgpa": 9.1},
    {"name": "Dev", "cgpa": 7.8}
]

sorted_students = sorted(
    students,
    key=lambda x: x["cgpa"],
    reverse=True
)
print(sorted_students)
```
