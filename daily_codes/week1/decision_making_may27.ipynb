```python
# read purchase cost and selling cost from the user
# state profit or loss and how much
```

```python
purchase_cost = int(input("enter the purchase cost"))
selling_cost = int(input("Enter the selling cost"))

if selling_cost > purchase_cost:
    profit = selling_cost - purchase_cost
    print("Profit",profit)
elif purchase_cost > selling_cost:
  lose = purchase_cost - selling_cost
  print("Loss",loss)

else:
    print("not profit or loss")
```

```python
year = int(input("enter the year"))

if year % 4==0:
    if year % 100 ==0:
        if year % 400 ==0:
            print(year," is century and leap year")
        else:
            print(year,"is century but not leap year")
    else:
        print(year,"is leap year")
else:
    print(year," is not leap year")
```

```python
"""
menu driven ->
take a string form the user
menu -> upper, 2 lowercase, 3 replace a perticular part of the string
for third opetion it should which part to replace with what

""";
```

```python
print("Welcome")

text = input("Enter a string")

print("menu=>\n1. uppercase\n2. lowercase\n3. replace part of string")

ch = int(input("select our choice"))

if ch == 1:
   print(text.upper())

elif ch == 2:
   print(text.lower())

elif ch == 3:
    r1 = input("Enter the part of string to replace:")
    r2 = input("Enter the word:")

    print(text.replace(r1,r2))

else:
   print("Invalid choice")
```
