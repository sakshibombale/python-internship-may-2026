```python
# write a program to find factorial of numbers using 5 
```

```python
num = 5 
fact = 1 

while num >= 1:
    fact = fact * num
    num -= 1

print("factorial",fact)
    
```

```python
# write a program to find factorial of numbers using user input 
```

```python
num = int(input("Enter the factorial of a number")) 
fact = 1 

while num >= 1:
    fact = fact * num
    num -= 1

print("factorial",fact)
```

```python
# find the addition of digits of a  numbers
# eg -> 12345 -> 1 + 2 + 3 + 4 + 5 = 15
```

```python
# hint -> make use of % and //
```

```python
num = 12345
sum = 0

while num > 0:
    rem = num % 10  # rem = remainder
    sum = sum + rem
    num = num // 10
    
print("Addition of numbers :",sum)
```
