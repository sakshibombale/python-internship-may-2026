```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
x = np.array([10, 23, 29, 34, 39, 45, 50])
y = np.array([14, 21, 31, 43, 48, 55, 58])
z = np.array([56, 50, 44, 37, 26, 20, 16])
```

```python
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("line chart: x vs y")
plt.plot(x, y, marker = 'o', mfc= 'red', ls = '--');
plt.grid();
# o, *, H, <, >, ^
```

```python
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("line chart: x vs y")
plt.plot(x, z, marker = 'o', mfc= 'orange', ls = '--');
plt.grid();
# o, *, H, <, >, ^
```

```python
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("line chart: x vs y")
plt.plot(x, y, marker = 'o', mfc= 'red', ls = '--', label = '2024');
plt.plot(x, z, marker = 'o', mfc= 'black', ls = '--', color = 'orange', label = '2025');
plt.grid();
plt.legend();
# o, *, H, <, >, ^
```

```python
plt.subplot(2, 2, 1)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Line Chart: x vs z")
plt.plot(x, y, marker='o', mfc='red', ls='--')
plt.grid();

plt.subplot(2, 2, 2)
plt.xlabel("x values")
plt.ylabel("z values")
plt.title("Line Chart: x vs z")
plt.plot(x, z, marker='o', mfc='red', ls='--')
plt.grid();
```

```python
plt.figure(figsize=(16, 9))

plt.subplot(2, 2, 1)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Line Chart: x vs z")
plt.plot(x, y, marker='o', mfc='red', ls='--')
plt.grid();

plt.subplot(2, 2, 2)
plt.xlabel("x values")
plt.ylabel("z values")
plt.title("Line Chart: x vs z")
plt.plot(x, z, marker='o', mfc='red', ls='--')
plt.grid();
```

```python
# load Social_Network_Ads.csv
# plot bar graph showing the count of how many people purchased and how many did 
# not 
# purchased column => 0 -> purchased, 1 -> not purchased
```

```python
ads = pd.read_csv("Social_Network_Ads.csv")
```

```python
ads
```

```python
ads["Purchased"].value_counts()
```

```python
ads["Purchased"].value_counts().index
```

```python
counts = ads["Purchased"].value_counts()
labels = ["Purchased", "Not Purchased"]

plt.bar(labels, [counts[0], counts[1]])

plt.xlabel("Purchase Status")
plt.ylabel("Number of People")
plt.title("Purchased vs Not Purchased")

plt.grid()
```
