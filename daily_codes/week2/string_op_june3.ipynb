```python
"""
take a string as user input
use a loop and if - else to check if a specific character is present in  the string or not.
get output for both if character is present and if it is not present.
take 2 characters to check  in the string

user input -> string, character1, character2
""";
```

```python
s = input("Enter a string: ")
ch1 = input("Enter first character: ")
ch2 = input("Enter second character: ")

found_ch1 = False
found_ch2 = False

for ch in s:
    if ch == ch1:
        found1 = True
    if ch == ch2:
        found2 = True

if found_ch1:
    print(ch1, "is present in the string")
else:
    print(ch1, "is not present in the string")

if found_ch2:
    print(ch2, "is present in the string")
else:
    print(ch2, "is not present in the string")
```
