```python
# create a new list with names which starts with 's' or 'S'
```

```python
names = ['Aryan', 'atharv', 'sammruddhi', 'swaraj', 'sukanya', 'shivraj']

new_list = [name for name in names if name.startswith('s') or name.startswith('S')]

print(new_list)
```

```python
for i in names :
    if i.lower().startswith('s'):
        print(i)
```

```python
new = [i for i in names if i.lower().startswith('s')]
new
```
