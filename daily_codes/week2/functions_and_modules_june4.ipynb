```python
def num(x,y):
    if x > y:
        print(x, "is larger")
    else:
        print(y, "is larger")
    
```

```python
num(23,56)
```

```python
def compare(x,y):
    if x > y:
        print(x, "is larger")
    elif y > x:
        print(y, "is larger")
    else:
        print("both are equal")
    
```

```python
compare(45,78)
```

```python
def factorial(num):
    fact = 1

    for i in range(1, num + 1):
        fact = fact * i
    print(fact)
```

```python
factorial(5)
```

```python
factorial(3)
```

```python
def addall(*args):
    s = 0
    for i in args:
        s = s + i
    return s
```

```python
# type hinting => tell the type of parameters and return value, and docstring 
```

```python
def sq(x: int) -> int:
    
    """
    calculate the square of a number.

    Args:
        x (int): The number to square.

    Returns:
        int: The square of the number
    """
    res = x*x
    return res
```

```python
help(sq)
```
