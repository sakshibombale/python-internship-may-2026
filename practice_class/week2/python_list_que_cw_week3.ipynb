```python
"""
Question 2: Sum of Even Numbers
Write Python code that iterates through the following list numbers = [12, 7, 23, 48, 15, 66, 9].
Calculate and print the sum of all even numbers in the list.

""";
```

```python
numbers = [12, 7, 23, 48, 15, 66, 9]

sum_even = 0

for num in numbers:
    if num % 2 == 0:
        sum_even += num

print("Sum of even numbers:", sum_even)
```

```python
"""
Question 3: Manual List Reversal
Given a list data = ['apple', 'banana', 'cherry', 'date', 'elderberry'],
create a new list that contains the elements of data in reverse order.
You are not allowed to use data.reverse() or data[::-1]. 
You must achieve this using loops and basic list operations.

""";
```

```python
data = ['apple', 'banana', 'cherry', 'date', 'elderberry']

reversed_list = []

for item in data:
    reversed_list.insert(0, item)

print("Reversed List:", reversed_list)
```

```python
"""
Question 4: Finding the Maximum Element
Without using the built-in max() function, write code to find and print the 
largest number in the list values = [8, 2, 10, 5, 17, 3, 14].

""";
```

```python
values = [8, 2, 10, 5, 17, 3, 14]

largest = values[0]

for num in values:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
"""
Question 5: Count Occurrences
Given the list items = ['red', 'blue', 'green', 'red', 'yellow', 'blue', 'red'],
write code to count how many times the string 'red' appears in the list and print the count. 
Do not use the list.count() method.

""";
```

```python
items = ['red', 'blue', 'green', 'red', 'yellow', 'blue', 'red']

count = 0

for color in items:
    if color == 'red':
        count += 1

print("Red appears", count, "times")
```

```python
"""
Question 6: Remove All Occurrences of an Element
Given a list mixed_numbers = [1, 5, 2, 5, 3, 5, 4, 5], write code to remove all instances of the number 5 from the list.
The final list should only contain [1, 2, 3, 4].
You must achieve this by iterating and creating a new list or modifying the existing one carefully.

""";
```

```python
mixed_numbers = [1, 5, 2, 5, 3, 5, 4, 5]

new_list = []

for num in mixed_numbers:
    if num != 5:
        new_list.append(num)

print(new_list)
```

```python
"""
Question 7: List Palindrome Check
Write code to determine if a given list is a palindrome (reads the same forwards and backwards).
Print "Palindrome" if it is, and "Not a Palindrome" otherwise.
Test with:
list1 = [1, 2, 3, 2, 1]
list2 = ['a', 'b', 'c', 'b', 'd']
You are not allowed to use [::-1] for comparison.

""";
```

```python
list1 = [1, 2, 3, 2, 1]
list2 = ['a', 'b', 'c', 'b', 'd']

for lst in [list1, list2]:
    reverse_list = []

    for item in lst:
        reverse_list.insert(0, item)

    if lst == reverse_list:
        print("Palindrome")
    else:
        print("Not a Palindrome")
```

```python
"""
Question 9: Flatten a Nested List
You are given a list that contains integers and other lists (only one level of nesting). For
example: nested_list = [1, [2, 3], 4, [5, 6, 7], 8].
Write code to create a new flat list containing all the integers from nested_list. 
The result should be [1, 2, 3, 4, 5, 6, 7, 8].

""";
```

```python
nested_list = [1, [2, 3], 4, [5, 6, 7], 8]

flat_list = []

for item in nested_list:
    if type(item) == list:
        for num in item:
            flat_list.append(num)
    else:
        flat_list.append(item)

print(flat_list)
```

```python
"""
Question 10: Find Common Elements
Given two lists, list_p = [1, 2, 3, 4, 5] and q = [4, 5, 6, 7, 8] write code to find and 
print all elements that are common to both lists.
The order of common elements does not matter. You are not allowed to use set operations.

""";
```

```python
list_p = [1, 2, 3, 4, 5]
list_q = [4, 5, 6, 7, 8]

common = []

for item in list_p:
    if item in list_q:
        common.append(item)

print("Common elements:", common)
```
